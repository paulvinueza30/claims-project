```python
import pandas as pd
```

```python
states = pd.read_csv(
    'https://www2.census.gov/geo/docs/reference/codes2020/national_state2020.txt',
    sep='|', dtype=str
)
states = states[['STATE_NAME', 'STATEFP']]
states.columns = [ 'state_name', 'state_fp']
```

```python
benef = pd.read_csv("./data/DE1_0_2008_Beneficiary_Summary_File_Sample_1.csv")
benef.columns

```

```python
benef_cols = {
    'DESYNPUF_ID': 'bene_id',
    'BENE_BIRTH_DT': 'birth_dt',
    'BENE_DEATH_DT': 'death_dt',
    'BENE_SEX_IDENT_CD': 'sex',
    'BENE_RACE_CD': 'race',
    'BENE_ESRD_IND': 'esrd',
    'SP_STATE_CODE': 'state',
    'BENE_COUNTY_CD': 'county',
    'BENE_HI_CVRAGE_TOT_MONS': 'hi_mos',
    'BENE_SMI_CVRAGE_TOT_MONS': 'smi_mos',
    'BENE_HMO_CVRAGE_TOT_MONS': 'hmo_mos',
    'PLAN_CVRG_MOS_NUM': 'plan_mos',
    'SP_ALZHDMTA': 'alzheimer',
    'SP_CHF': 'chf',
    'SP_CHRNKIDN': 'ckd',
    'SP_CNCR': 'cancer',
    'SP_COPD': 'copd',
    'SP_DEPRESSN': 'depression',
    'SP_DIABETES': 'diabetes',
    'SP_ISCHMCHT': 'ischemic_heart',
    'SP_OSTEOPRS': 'osteoporosis',
    'SP_RA_OA': 'arthritis',
    'SP_STRKETIA': 'stroke_tia',
    'MEDREIMB_IP': 'ip_reimb',
    'BENRES_IP': 'ip_bene_resp',
    'MEDREIMB_CAR': 'car_reimb',
    'BENRES_CAR': 'car_bene_resp',
    'PPPYMT_CAR': 'car_ppp_pmt',
}
benef.rename(columns=benef_cols, inplace=True)
```

```python
print(benef.columns)
states.head(3)
```

```python
benef['sex'] = benef['sex'].map({1:'M', 2:'F'})
benef['state'] = benef['state'].astype(str)
benef = benef.merge(states, left_on='state', right_on='state_fp', how='left')
benef = benef.drop(['state_fp', 'county', 'race' ],axis=1)
benef.head(3)
```
